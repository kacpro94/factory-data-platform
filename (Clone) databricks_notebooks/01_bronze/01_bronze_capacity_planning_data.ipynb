```python
%sql
CREATE OR REPLACE TABLE data_warehouse_factory.bronze.capacity_planner_events_bronze AS
    SELECT
        CAST(ID AS INT) AS id,
        CAST(Employee AS STRING) AS employee_full_name,
        ABS(xxhash64(LOWER(TRIM(Employee)))) AS employee_id,
        ABS(xxhash64(TRIM(`Event Type`))) AS event_type_id,
        CAST(`Event Type` AS STRING) AS event_type_name,
        CAST(`Swap cell` AS STRING) AS cell_name_swap,
        CAST(try_to_timestamp(`Start Datetime`, 'M/d/yyyy h:mm a') AS DATE) AS start_date,
        CAST(try_to_timestamp(`End Datetime`, 'M/d/yyyy h:mm a') AS DATE) AS end_date,
        DATE_FORMAT(try_to_timestamp(`Start Datetime`, 'M/d/yyyy h:mm a'), 'HH:mm:ss') AS start_time,
        DATE_FORMAT(try_to_timestamp(`End Datetime`, 'M/d/yyyy h:mm a'), 'HH:mm:ss') AS end_time,
        try_to_timestamp(`Start Datetime`, 'M/d/yyyy h:mm a') AS start_datetime,
        try_to_timestamp(`End Datetime`, 'M/d/yyyy h:mm a') AS end_datetime,
        CAST(`Remarks` AS STRING) AS remarks,
        TRY_CAST(`FTE` AS INT) AS fte
FROM read_files('/Volumes/data_warehouse_factory/bronze/capacity_planing_data_raw/capacity_planing_data_raw.csv',format => 'csv', header => 'true')
```
