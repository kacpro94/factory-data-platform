```python
%sql
CREATE OR REPLACE TABLE data_warehouse_factory.gold.dim_production_cells AS
SELECT 
  cell_key,
  
  -- Kody i nazwy linii
  upper(trim(line_code)) AS line_code,
  trim(line_name) AS line_name,
  
  -- Kody i nazwy komórek
  upper(trim(cell_code)) AS cell_code,
  trim(cell_name) AS cell_name,
  
  -- Przydatna etykieta złożona do wizualizacji w Power BI
  concat(trim(line_name), ' - ', trim(cell_name)) AS line_cell_name,
  concat(trim(cell_name), ' (', upper(trim(cell_code)), ')') AS cell_display_label,
  
  record_type,
  _silver_ingested_at,
  CURRENT_TIMESTAMP() AS _gold_created_at
FROM data_warehouse_factory.silver.silver_production_structure;
```
