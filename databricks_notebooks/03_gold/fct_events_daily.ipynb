```python
%sql
CREATE OR REPLACE TABLE data_warehouse_factory.gold.fct_events_daily AS
SELECT 
  e.daily_event_key,
  e.parent_event_key,
  CAST(date_format(e.production_date, 'yyyyMMdd') AS INT) AS date_key,
  CAST(e.employee_key AS INT) AS employee_key,
  c.cell_key,
  upper(trim(e.swap_cell)) AS cell_code,
  trim(e.employee_full_name) AS employee_full_name,
  trim(e.event_type) AS event_type,
  e.start_date_general,
  e.end_date_general,
  e.production_date,
  date_format(e.daily_event_start, 'HH:mm:ss') AS start_time,
  date_format(e.daily_event_end, 'HH:mm:ss') AS end_time,
  e.daily_event_start,
  e.daily_event_end,
  e.duration_hours,
  e.duration_minutes,
  e.fte,
  e.remarks,
  e.source,
  e._silver_ingested_at,
  CURRENT_TIMESTAMP() AS _gold_created_at
FROM data_warehouse_factory.silver.silver_events_daily e
LEFT JOIN data_warehouse_factory.gold.dim_production_cells c
  ON upper(trim(e.swap_cell)) = c.cell_code;
```
